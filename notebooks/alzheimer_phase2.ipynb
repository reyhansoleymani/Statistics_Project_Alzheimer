## آماده‌سازی محیط و بارگذاری داده‌ها


```python
import zipfile
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

sns.set_theme(style="whitegrid", context="notebook")

PROJECT_DIR = Path.cwd()
DATA_DIR = PROJECT_DIR / "data"
ZIP_PATH = DATA_DIR / "alzheimer.zip"
EXTRACT_DIR = DATA_DIR / "extracted"

EXTRACT_DIR.mkdir(parents=True, exist_ok=True)

uds_matches = list(EXTRACT_DIR.rglob("udslong_imputed.csv"))
mri_matches = list(EXTRACT_DIR.rglob("mriscanclean.csv"))

if not (uds_matches and mri_matches):
    if not ZIP_PATH.exists():
        raise FileNotFoundError(
            f"Could not find {ZIP_PATH}. Put alzheimer.zip inside the data folder "
            "or update ZIP_PATH to the correct location."
        )
    with zipfile.ZipFile(ZIP_PATH, "r") as zip_ref:
        zip_ref.extractall(EXTRACT_DIR)

uds_path = next(EXTRACT_DIR.rglob("udslong_imputed.csv"), None)
mri_path = next(EXTRACT_DIR.rglob("mriscanclean.csv"), None)

if uds_path is None or mri_path is None:
    raise FileNotFoundError("The required UDS and/or MRI CSV file was not found.")

uds = pd.read_csv(uds_path)
mri = pd.read_csv(mri_path)

print("UDS file:", uds_path.name)
print("MRI file:", mri_path.name)
print("UDS shape:", uds.shape)
print("MRI shape:", mri.shape)
```

## احتمال حاشیه ای و مشترک:

```python
# انتخاب متغیرهای موردنیاز از دیتاست کلین‌شده
df = combined1[["NACCAGE", "RH_ENTORHINAL_AVGTH"]].copy()

# حذف مشاهداتی که برای یکی از این دو متغیر مقدار گمشده دارند
df = df.dropna(subset=["NACCAGE", "RH_ENTORHINAL_AVGTH"])

# محاسبه میانه ضخامت قشر انتورینال راست
median_thickness = df["RH_ENTORHINAL_AVGTH"].median()

# تعریف رویداد A: سن حداقل 75 سال
A = df["NACCAGE"] >= 75

# تعریف رویداد B: ضخامت کمتر از میانه نمونه
B = df["RH_ENTORHINAL_AVGTH"] < median_thickness

# تعداد کل مشاهدات معتبر
n = len(df)

# تعداد وقوع رویدادهای A و B و اشتراک آن‌ها
n_A = A.sum()
n_B = B.sum()
n_AB = (A & B).sum()

# محاسبه احتمال‌های تجربی
P_C_A = n_A / n
P_C_B = n_B / n
P_C_AB = n_AB / n

# نمایش نتایج
print("Total observations (n):", n)
print("Median of right entorhinal thickness:", median_thickness)

print("\nEvent A:")
print("n_A =", n_A)
print("P_C(A) =", round(P_C_A, 4))

print("\nEvent B:")
print("n_B =", n_B)
print("P_C(B) =", round(P_C_B, 4))

print("\nIntersection AB:")
print("n_AB =", n_AB)
print("P_C(AB) =", round(P_C_AB, 4))

print("\nProbabilities in percentage:")
print("P_C(A) =", round(P_C_A * 100, 2), "%")
print("P_C(B) =", round(P_C_B * 100, 2), "%")
print("P_C(AB) =", round(P_C_AB * 100, 2), "%")
```

## بررسی مفهوم استقلال : 


```python
import pandas as pd

# انتخاب متغیرهای لازم
df = combined1[["NACCAGE", "SEX"]].copy()

# تبدیل به عددی و حذف مقادیر گمشده
df["NACCAGE"] = pd.to_numeric(df["NACCAGE"], errors="coerce")
df["SEX"] = pd.to_numeric(df["SEX"], errors="coerce")
df = df.dropna(subset=["NACCAGE", "SEX"])

# حذف کدهای نامعتبر یا ناشناخته جنس
df = df[df["SEX"].isin([1, 2])]

# تعریف رویدادها
C = df["NACCAGE"] >= 75
E = df["SEX"] == 1

# محاسبه تعدادها
N = len(df)
nC = C.sum()
nE = E.sum()
nCE = (C & E).sum()

# احتمال‌های تجربی
P_C = nC / N
P_E = nE / N
P_CE = nCE / N

# بررسی استقلال
P_product = P_C * P_E

print("N =", N)
print("nC =", nC)
print("nE =", nE)
print("nCE =", nCE)

print("\nP(C) =", round(P_C, 4))
print("P(E) =", round(P_E, 4))
print("P(CE) =", round(P_CE, 4))
print("P(C) * P(E) =", round(P_product, 4))

print("\nDifference =", round(abs(P_CE - P_product), 4))
```

## توزیع گسسته و مدل دوجمله ای: 


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# استخراج جنسیت از دیتافریم واقعی
df = combined1[["SEX"]].copy()
df["SEX"] = pd.to_numeric(df["SEX"], errors="coerce")
df = df[df["SEX"].isin([1, 2])]

# پارامترها
N = len(df)
K = (df["SEX"] == 1).sum()
p = K / N

n = 10
B = 10000
rng = np.random.default_rng(42)

# نمونه گیری تصادفی 10 نفره بدون جایگذاری
sex = df["SEX"].to_numpy()

X_sim = np.array([
    (rng.choice(sex, size=n, replace=False) == 1).sum()
    for _ in range(B)
])

# احتمال تجربی
k = np.arange(n + 1)
P_empirical = np.array([
    np.mean(X_sim == i) for i in k
])

# احتمال نظری دوجمله ای
P_binomial = binom.pmf(k, n, p)

# نمودار
plt.figure(figsize=(9, 5))

plt.bar(k - 0.2, P_empirical, width=0.4, label="Empirical")
plt.bar(k + 0.2, P_binomial, width=0.4, label="Binomial")

plt.xlabel("Number of men in sample of 10")
plt.ylabel("Probability")
plt.title("Empirical vs Binomial Distribution")

plt.xticks(k)
plt.legend(loc="upper right")  # محل ثابت، بدون جستجوی loc="best"
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("binomial_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 1. RH_ENTORHINAL_AVGTH

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# --------------------------------
# 1. آماده‌سازی داده‌ها
# --------------------------------

column = "RH_ENTORHINAL_AVGTH"

x = pd.to_numeric(combined1[column], errors="coerce")
x = x.replace([np.inf, -np.inf], np.nan).dropna()

# --------------------------------
# 2. آمار توصیفی
# --------------------------------

n = len(x)
mean = x.mean()
std = x.std(ddof=1)
median = x.median()
minimum = x.min()
maximum = x.max()
q1 = x.quantile(0.25)
q3 = x.quantile(0.75)

print("Descriptive Statistics")
print("----------------------")
print("Number of observations:", n)
print("Mean:", round(mean, 4))
print("Standard deviation:", round(std, 4))
print("Median:", round(median, 4))
print("Minimum:", round(minimum, 4))
print("Maximum:", round(maximum, 4))
print("First quartile (Q1):", round(q1, 4))
print("Third quartile (Q3):", round(q3, 4))

# --------------------------------
# 3. آزمون Shapiro-Wilk
# --------------------------------

W, p_value = stats.shapiro(x)

alpha = 0.05

print("\nShapiro-Wilk Normality Test")
print("---------------------------")
print("W statistic:", round(W, 5))
print("p-value:", p_value)
print("Significance level:", alpha)

if p_value < alpha:
    print("Conclusion: Reject H0. There is evidence against normality.")
else:
    print("Conclusion: Do not reject H0. There is insufficient evidence against normality.")

# --------------------------------
# 4. محاسبه چولگی و کشیدگی
# --------------------------------

skewness = stats.skew(x)
kurtosis = stats.kurtosis(x)

print("\nDistribution Shape")
print("------------------")
print("Skewness:", round(skewness, 4))
print("Excess Kurtosis:", round(kurtosis, 4))

# --------------------------------
# 5. رسم هیستوگرام و Q-Q Plot
# --------------------------------

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram + fitted normal curve
axes[0].hist(
    x,
    bins=30,
    density=True,
    alpha=0.7,
    edgecolor="black",
    label="Observed data"
)

xx = np.linspace(x.min(), x.max(), 300)

normal_curve = stats.norm.pdf(
    xx,
    loc=mean,
    scale=std
)

axes[0].plot(
    xx,
    normal_curve,
    linewidth=2,
    label="Fitted normal distribution"
)

axes[0].set_title("Histogram of Right Entorhinal Thickness")
axes[0].set_xlabel("Right Entorhinal Average Thickness")
axes[0].set_ylabel("Density")
axes[0].legend(loc="upper right")
axes[0].grid(alpha=0.3)

# Q-Q Plot
stats.probplot(x, dist="norm", plot=axes[1])

axes[1].set_title("Q-Q Plot")
axes[1].set_xlabel("Theoretical Quantiles")
axes[1].set_ylabel("Observed Quantiles")
axes[1].grid(alpha=0.3)

plt.tight_layout()

# ذخیره نمودار برای گزارش
plt.savefig(
    "task6_normality_plots.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## احتمال شرطی وضعیت شناختی: 

```python
import pandas as pd

data = combined1

# انتخاب متغیرهای مورد نیاز و حذف داده‌های ناقص
valid = data[["MOCATOTS", "RH_ENTORHINAL_AVGTH"]].dropna()

# محاسبه میانه‌ها
cognitive_median = valid["MOCATOTS"].median()
mri_median = valid["RH_ENTORHINAL_AVGTH"].median()

# تعریف وضعیت شناختی پایین
valid["Low_Cognition"] = valid["MOCATOTS"] <= cognitive_median

# گروه‌بندی بر اساس MRI
valid["MRI_Group"] = valid["RH_ENTORHINAL_AVGTH"].apply(
    lambda x: "Low MRI Thickness" if x < mri_median
    else "High MRI Thickness"
)

# محاسبه احتمال شرطی در هر گروه
result = valid.groupby("MRI_Group")["Low_Cognition"].agg(
    ["sum", "count", "mean"]
)

result.columns = [
    "Number with Low Cognition",
    "Group Size",
    "Conditional Probability"
]

print("Cognitive median:", cognitive_median)
print("MRI median:", mri_median)
print(result)

# احتمال کلی نمره شناختی پایین
print("Overall probability:",
      valid["Low_Cognition"].mean())
```
